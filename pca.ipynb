# Principal Component Analysis of Grimm Brothers

## Setup and Imports

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import normalize
from sklearn.decomposition import PCA, TruncatedSVD as SVD
```

```python
import plotly.express as px
import plotly.io as pio
```

```python
TFIDF = pd.read_csv('data/pg2591-TFIDF.csv',index_col=0)
TFIDF.head()
```

```python
DTM = pd.read_csv('data/pg2591-DTM.csv',index_col=0)
DTM.head()
```

```python
LIB = pd.read_csv('data/pg2591-LIB.csv',index_col=0)
LIB.head()
```

## DCM

```python
pca_engine = PCA(n_components=5)
DCM = pd.DataFrame(pca_engine.fit_transform(TFIDF), index=TFIDF.index)
DCM.columns = ['PC{}'.format(i) for i in DCM.columns]
DCM
```

## Make Loading

```python
LOADING = pd.DataFrame(pca_engine.components_.T * np.sqrt(pca_engine.explained_variance_))
LOADING.columns = ["PC{}".format(i) for i in LOADING.columns]
LOADING.index = TFIDF.columns
LOADING.index.name = 'term_str'
LOADING.head()
```

## Get Top 5 Postive Terms for PC0

```python
LOADING.sort_values(by='PC0',ascending=False).head(5)
```

## Get Top 5 Negative Terms for PC1

```python
LOADING.sort_values(by='PC1',ascending=True).head(5)
```

## COMPS

```python
# Got help from peer for this code
COMPS= pd.DataFrame(columns=['PC0_pole_1','PC0_pole_2','PC1_pole_1','PC1_pole_2','PC2_pole_1','PC2_pole_2','PC3_pole_1','PC3_pole_2','PC4_pole_1','PC4_pole_2'])
COMPS['PC0_pole_1'] = LOADING.sort_values(by='PC0',ascending=True).head(5).index.to_list()
COMPS['PC0_pole_2'] = LOADING.sort_values(by='PC0',ascending=True).tail(5).index.to_list()
COMPS['PC1_pole_1'] = LOADING.sort_values(by='PC1',ascending=True).head(5).index.to_list()
COMPS['PC1_pole_2'] = LOADING.sort_values(by='PC1',ascending=True).tail(5).index.to_list()
COMPS['PC2_pole_1'] = LOADING.sort_values(by='PC2',ascending=True).head(5).index.to_list()
COMPS['PC2_pole_2'] = LOADING.sort_values(by='PC1',ascending=True).tail(5).index.to_list()
COMPS['PC3_pole_1'] = LOADING.sort_values(by='PC3',ascending=True).head(5).index.to_list()
COMPS['PC3_pole_2'] = LOADING.sort_values(by='PC3',ascending=True).tail(5).index.to_list()
COMPS['PC4_pole_1'] = LOADING.sort_values(by='PC4',ascending=True).head(5).index.to_list()
COMPS['PC4_pole_2'] = LOADING.sort_values(by='PC4',ascending=True).tail(5).index.to_list()
COMPS
```

## Creating Visuals

### Getting DOC Table ready

```python
DOC = DTM.sum(1).to_frame('n_tokens')
DOC['n_types'] = DTM.astype('bool').sum(1)
DOC['pkr'] = DOC.n_types / DOC.n_tokens
DOC = DOC.join(LIB['volume'])
```

```python
DOC.head()
```

```python
NEW=DCM.join(DOC)
NEW
```

### Importing Functions

```python
def vis_pcs(a, b,DCM, col='doc_label'):
    fig =px.scatter(DCM, 
        f"PC{a}", f"PC{b}", 
        color=DOC[col].astype('category'), 
        hover_name=DCM.index,
        marginal_x='box', 
        height=1000, 
        width=1200)
    return fig
```

```python
def vis_loadings(COMPS, a=0, b=1, hover_name='term_str'):
    return px.scatter(COMPS.reset_index(), f"PC{a}", f"PC{b}", 
                      text='term_str', 
                      marginal_x='box', 
                      height=800)
```

```python
DOC.head()
```

### Visualize PCA 0 and 1

```python
fig1=vis_pcs(0,1,NEW,col='volume')
fig1.write_image('output/p2591-PC1.png')
fig1.show()
```

The new additions in Volume II are all on the negative side for PC0 and PC1. Volume I is similar but also has stories that has more positive componets. This suggests that Volume I is more varied in PC0 and PC1 than Volume II.

### Visualize Loadings 0 and 1

```python
fig2=vis_loadings(LOADING,0,1)
fig2.write_image('output/p2591-LOADING1.png')
fig2.show()
```

Most words seems to be in a huge cluster around the middle. There are some other clusters. For example the bottom left seems to have royal words like prince and kings. Right above the cluster seems to have family/identity words like mother and children.

## VIS PCA 2,3

```python
fig3=vis_pcs(2,3,NEW,col='volume')
fig3.write_image('output/p2591-PC2.png')
fig3.show()
```

Both Volume I and Volume II seem to be well to be evenly distriuted around PC2 and PC3 with even spread in the postive and negative side. Again Volume II is less varied - as it has a shorter whiskser - with PC2 and PC3 values, this could be due to Volume II having less stories

## VIS LOADINGS 2,3

```python
fig4=vis_loadings(LOADING,2,3)
fig4.write_image('output/p2591-LOADING2.png')
fig4.show()
```

Like the last loading, this one also have a huge clusters around the origin and several smaller clusters. This one seems to more smaller custers than the loadings of PC0 and PC1. I did notice similar separatuion class. Working class terms - cookm huntsmen, etc -have negative PC3 value, while prince and princess - royal words - are on the postive side of PC3. There are clear outliers though with king and queen having negative PC3. SO maybe these PC are a bit more complex than just social class.
